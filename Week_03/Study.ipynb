```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t=np.linspace(0, 10*np.pi,1000)
w_0=1
K = np.array([ i for i in range(1, 10)])

y = 0
y_2 = 0
y_3 = 0
for i in range(len(K)):
    y += np.sin(K[i]*w_0*t)+ np.cos(K[i]*w_0*t)
    y_2 += np.cos(K[i]*w_0*t)*np.cos(K[i]*w_0*t)
    y_3 += np.sin(K[i]*w_0*t)*np.cos(K[i]*w_0*t)

#y=np.sin(K*w_0*t)*np.sin(K*w_0*t)
# y_2=np.cos(K*w_0*t)*np.cos(K*w_0*t)
# y_3=np.sin(K*w_0*t)*np.cos(K*w_0*t)
```

```python
plt.plot(t,y,label='sin(K*w_0*t)')
# plt.plot(t,y_2,label='cos(K*w_0*t)')
# plt.plot(t,y_3,label='sin(K*w_0*t)*cos(K*w_0*t)')
plt.xlabel('t')
plt.ylabel('y')
plt.legend()
plt.show()
```
